```python
import pandas as pd
import matplotlib.pyplot as plt
import io

# Step 1: Prepare the data
# I have embedded all the data we prepared into a single CSV string.
csv_data = """model,dataset,Intervention,value,Acc@1,Acc@3,Acc@5,Acc@10,F1,MRR
MHSA,EPR,P,0.1,12.366,34.721,44.485,55.291,10.310,26.795
MHSA,EPR,P,0.25,10.215,22.638,29.313,38.253,9.198,19.437
MHSA,EPR,P,0.5,16.323,25.246,29.164,34.449,16.180,22.695
MHSA,EPR,P,0.75,23.829,34.624,38.559,43.205,23.641,30.821
MHSA,EPR,P,0.9,26.158,38.878,43.610,49.376,25.875,34.395
MHSA,EPR,gamma,0.1,13.502,22.429,26.818,32.890,13.311,20.150
MHSA,EPR,gamma,0.25,8.247,18.263,23.995,32.099,7.530,16.126
MHSA,EPR,gamma,0.5,7.330,21.182,29.272,40.131,6.043,17.851
MHSA,EPR,gamma,0.75,8.440,25.886,35.690,47.547,6.765,20.992
MHSA,EPR,gamma,0.9,9.478,29.162,39.721,52.186,7.660,23.247
MHSA,EPR,rho,0.1,26.087,49.595,57.699,65.897,23.527,40.379
MHSA,EPR,rho,0.25,13.119,30.416,38.382,48.027,11.337,24.898
MHSA,EPR,rho,0.5,9.190,20.451,26.607,34.952,8.428,17.700
MHSA,EPR,rho,0.75,10.617,19.000,23.409,29.747,10.347,17.115
MHSA,EPR,rho,0.9,12.030,19.710,23.438,28.824,11.855,17.875
Mamba,EPR,P,0.1,13.804,34.764,43.634,53.761,11.136,27.390
Mamba,EPR,P,0.25,9.815,21.915,28.453,37.220,8.244,18.875
Mamba,EPR,P,0.5,14.674,23.593,27.729,33.364,14.392,21.171
Mamba,EPR,P,0.75,23.533,34.273,38.254,43.205,23.089,30.575
Mamba,EPR,P,0.9,27.589,40.102,44.698,50.403,26.988,35.730
Mamba,EPR,gamma,0.1,12.402,21.250,25.708,31.938,12.034,19.062
Mamba,EPR,gamma,0.25,7.812,17.542,23.064,31.075,6.677,15.509
Mamba,EPR,gamma,0.5,8.086,21.379,29.089,39.383,6.235,18.208
Mamba,EPR,gamma,0.75,9.370,25.797,35.042,46.312,7.428,21.314
Mamba,EPR,gamma,0.9,10.900,29.313,38.986,50.700,8.450,23.839
Mamba,EPR,rho,0.1,22.927,44.930,52.342,60.907,20.510,36.487
Mamba,EPR,rho,0.25,13.534,29.485,37.262,46.675,11.091,24.584
Mamba,EPR,rho,0.5,8.777,19.727,25.757,34.025,7.519,17.124
Mamba,EPR,rho,0.75,9.448,17.721,22.297,28.811,8.915,15.970
Mamba,EPR,rho,0.9,10.725,18.413,22.283,27.854,10.406,16.633
LSTM,EPR,P,0.1,15.620,35.344,43.313,52.188,11.511,28.273
LSTM,EPR,P,0.25,9.363,20.680,26.834,34.963,6.480,17.873
LSTM,EPR,P,0.5,6.921,14.168,18.398,24.553,5.938,12.829
LSTM,EPR,P,0.75,9.337,17.347,21.510,27.404,8.984,15.465
LSTM,EPR,P,0.9,11.250,20.445,25.280,32.003,10.921,18.255
LSTM,EPR,gamma,0.1,6.969,14.486,18.819,25.195,5.717,13.081
LSTM,EPR,gamma,0.25,7.326,16.571,21.758,29.213,5.033,14.606
LSTM,EPR,gamma,0.5,9.476,22.524,29.509,38.504,6.497,19.064
LSTM,EPR,gamma,0.75,11.081,27.035,35.336,45.368,7.937,22.313
LSTM,EPR,gamma,0.9,12.492,30.504,39.137,49.293,9.023,24.714
LSTM,EPR,rho,0.1,28.398,48.624,55.120,62.340,23.346,40.697
LSTM,EPR,rho,0.25,14.207,29.514,36.400,44.923,10.498,24.688
LSTM,EPR,rho,0.5,8.143,18.236,23.788,31.550,5.681,15.915
LSTM,EPR,rho,0.75,6.240,13.638,17.990,24.404,4.756,12.301
LSTM,EPR,rho,0.9,5.929,12.414,16.150,21.924,4.846,11.337
MHSA,DT-EPR,gamma,0.1,59.010,61.966,62.554,63.345,45.981,60.874
MHSA,DT-EPR,gamma,0.25,80.923,83.131,83.449,83.854,73.814,82.210
MHSA,DT-EPR,gamma,0.5,92.383,93.760,94.026,94.311,89.715,93.185
MHSA,DT-EPR,gamma,0.75,94.117,95.400,95.691,96.196,92.422,94.921
MHSA,DT-EPR,gamma,0.9,92.758,94.469,94.886,95.664,91.061,93.864
MHSA,DT-EPR,p,0.1,86.937,89.188,89.499,89.852,82.224,88.202
MHSA,DT-EPR,p,0.25,70.519,73.811,74.356,75.016,60.246,72.463
MHSA,DT-EPR,p,0.5,45.127,48.415,49.176,50.241,29.892,47.306
MHSA,DT-EPR,p,0.75,19.874,22.919,23.769,25.147,8.005,22.130
MHSA,DT-EPR,p,0.9,5.779,7.527,8.334,9.795,1.209,7.484
MHSA,DT-EPR,rho,0.1,72.238,77.129,78.514,80.920,65.555,75.431
MHSA,DT-EPR,rho,0.25,83.240,85.433,85.891,86.750,77.717,84.643
MHSA,DT-EPR,rho,0.5,77.065,79.541,79.939,80.410,68.736,78.525
MHSA,DT-EPR,rho,0.75,69.784,72.452,72.914,73.506,59.043,71.403
MHSA,DT-EPR,rho,0.9,64.698,67.334,67.838,68.495,52.617,66.339
Mamba,DT-EPR,gamma,0.1,59.050,61.932,62.406,63.056,45.737,60.835
Mamba,DT-EPR,gamma,0.25,80.927,83.085,83.396,83.775,73.775,82.188
Mamba,DT-EPR,gamma,0.5,92.351,93.716,94.000,94.272,89.652,93.144
Mamba,DT-EPR,gamma,0.75,93.902,95.155,95.429,95.957,92.042,94.686
Mamba,DT-EPR,gamma,0.9,92.335,94.140,94.596,95.331,90.026,93.484
Mamba,DT-EPR,p,0.1,86.881,89.115,89.463,89.821,82.222,88.151
Mamba,DT-EPR,p,0.25,70.564,73.746,74.253,74.858,60.243,72.445
Mamba,DT-EPR,p,0.5,45.126,48.322,48.892,49.775,29.801,47.191
Mamba,DT-EPR,p,0.75,20.232,23.016,23.584,24.530,7.967,22.213
Mamba,DT-EPR,p,0.9,6.496,8.055,8.534,9.429,1.160,7.896
Mamba,DT-EPR,rho,0.1,70.792,75.810,77.345,79.510,63.169,74.074
Mamba,DT-EPR,rho,0.25,83.072,85.317,85.768,86.393,77.283,84.451
Mamba,DT-EPR,rho,0.5,77.110,79.486,79.866,80.303,68.693,78.521
Mamba,DT-EPR,rho,0.75,69.819,72.427,72.824,73.353,58.984,71.383
Mamba,DT-EPR,rho,0.9,64.719,67.277,67.686,68.241,52.499,66.287
LSTM,DT-EPR,gamma,0.1,59.071,61.978,62.477,63.108,45.675,60.835
LSTM,DT-EPR,gamma,0.25,80.889,83.117,83.427,83.808,73.722,82.175
LSTM,DT-EPR,gamma,0.5,92.258,93.653,93.920,94.214,89.581,93.074
LSTM,DT-EPR,gamma,0.75,94.027,95.290,95.625,96.019,92.319,94.797
LSTM,DT-EPR,gamma,0.9,92.871,94.301,94.741,95.518,91.010,93.837
LSTM,DT-EPR,p,0.1,86.941,89.156,89.485,89.827,82.214,88.192
LSTM,DT-EPR,p,0.25,70.449,73.785,74.310,74.947,60.063,72.392
LSTM,DT-EPR,p,0.5,45.149,48.325,48.906,49.794,29.610,47.161
LSTM,DT-EPR,p,0.75,20.201,22.978,23.574,24.448,7.802,22.094
LSTM,DT-EPR,p,0.9,6.215,7.739,8.247,9.093,1.131,7.522
LSTM,DT-EPR,rho,0.1,71.088,76.393,78.139,80.690,63.472,74.529
LSTM,DT-EPR,rho,0.25,83.082,85.497,86.001,86.780,77.363,84.564
LSTM,DT-EPR,rho,0.5,77.094,79.522,79.907,80.365,68.634,78.522
LSTM,DT-EPR,rho,0.75,69.778,72.419,72.830,73.363,58.876,71.347
LSTM,DT-EPR,rho,0.9,64.699,67.301,67.734,68.288,52.383,66.270
"""

# Read the data into a pandas DataFrame
df = pd.read_csv(io.StringIO(csv_data))

# Step 2: Create the plotting function
def generate_comparison_plots(data, dataset_name):

    interventions = [('p', '$p^{new}$'), 
                     ('rho', '$\\mu_{\\rho}$ of $P(\\rho)$'), 
                     ('gamma', '$\\mu_{\\gamma}$ of $P(\\gamma)$')]
    
    models = data['model'].unique()
    
    # Define colors and markers for each model
    colors = {'MHSA': 'tab:blue', 'Mamba': 'tab:green', 'LSTM': 'tab:orange'}
    markers = {'MHSA': 'o', 'Mamba': 's', 'LSTM': '^'}

    # Create a 3x2 subplot grid
    fig, axes = plt.subplots(3, 2, figsize=(12, 15))
    fig.suptitle(f'Model Performance Comparison on {dataset_name} Dataset', fontsize=16, y=1.02)


    subplot_labels = [['(a1)', '(a2)'], ['(b1)', '(b2)'], ['(c1)', '(c2)']]

    for i, (intervention_code, intervention_label) in enumerate(interventions):
        # Filter data for the current intervention
        intervention_df = data[data['Intervention'] == intervention_code]
        
        ax_acc = axes[i, 0]
        for model in models:
            model_df = intervention_df[intervention_df['model'] == model].sort_values('value')
            ax_acc.plot(model_df['value'], model_df['Acc@1'], 
                        marker=markers.get(model, 'x'), 
                        color=colors.get(model, 'black'), 
                        label=model)
        
        ax_acc.set_xlabel(intervention_label, fontsize=12)
        ax_acc.set_ylabel('Acc@1 (%)', fontsize=12)
        ax_acc.legend()
        ax_acc.grid(True, linestyle='--', alpha=0.6)
        ax_acc.text(0.05, 0.1, subplot_labels[i][0], transform=ax_acc.transAxes, 
                    fontsize=18, fontweight='bold', va='bottom')

        ax_mrr = axes[i, 1]
        for model in models:
            model_df = intervention_df[intervention_df['model'] == model].sort_values('value')
            ax_mrr.plot(model_df['value'], model_df['MRR'], 
                        marker=markers.get(model, 'x'), 
                        color=colors.get(model, 'black'), 
                        label=model)
            
        ax_mrr.set_xlabel(intervention_label, fontsize=12)
        ax_mrr.set_ylabel('MRR (%)', fontsize=12)
        ax_mrr.legend()
        ax_mrr.grid(True, linestyle='--', alpha=0.6)
        ax_mrr.text(0.05, 0.1, subplot_labels[i][1], transform=ax_mrr.transAxes, 
                    fontsize=18, fontweight='bold', va='bottom')

    plt.tight_layout(rect=[0, 0, 1, 1]) # Adjust layout
    plt.show()


```

```python
df_epr = df[df['dataset'] == 'EPR']
generate_comparison_plots(df_epr, 'EPR')
```
