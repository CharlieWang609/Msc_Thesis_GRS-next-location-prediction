# Intervention on $\mu | \gamma$

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('gamma_intervented.csv')
df_melted = df.melt(id_vars=['mu_gamma', 'model'], value_vars=['F1', 'acc@1', 'mrr'],
                     var_name='metric', value_name='value')

metrics_to_plot = ['F1', 'acc@1', 'mrr']
for metric in metrics_to_plot:
    plt.figure(figsize=(10, 6))
    subset = df_melted[df_melted['metric'] == metric]
    for model_type in subset['model'].unique():
        model_subset = subset[subset['model'] == model_type]
        plt.plot(model_subset['mu_gamma'], model_subset['value'], marker='o',
                 label=f'{model_type}')

    plt.title(f'{metric} vs. $\mu | \gamma$ for Different Models')
    plt.xlabel('$\mu | \gamma$')
    plt.ylabel(f'{metric}')
    plt.grid(True)
    plt.legend(title='Model')
    plt.tight_layout()
    plt.show()
```
